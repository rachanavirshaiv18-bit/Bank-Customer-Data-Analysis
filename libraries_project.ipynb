```python

fake = Faker()

# Number of customers
num_customers = 200

# Create data
data = {
    "CustomerID": ["CUST" + str(i) for i in range(1001, 1001 + num_customers)],
    "Name": [fake.name() for i in range(num_customers)],
    "Age": np.random.randint(18, 70, num_customers),
    "Gender": np.random.choice(["Male", "Female"], num_customers),
    "Income": np.random.randint(20000, 100000, num_customers),
    "AccountBalance": np.random.randint(1000, 50000, num_customers)
}

# Create DataFrame
df = pd.DataFrame(data)

# Save to CSV
df.to_csv("bank_customers.csv", index=False)

df
```

```python
import pandas as pd
import matplotlib.pyplot as plt


# ==================== LOAD DATASET ====================

print("\n========== LOADING DATASET ==========")

# Read the CSV file
df = pd.read_csv("bank_customers.csv")

print("Dataset loaded successfully.")


# ==================== DATA CHECKING ====================

print("\n========== CHECKING DATA ==========")

# Display the first 5 rows
print("\nFirst 5 rows of the dataset:")
print(df.head())


# ==================== HANDLING MISSING VALUES ====================

print("\n========== HANDLING MISSING VALUES ==========")

# Check the number of missing values in each column
print(df.isnull().sum())


# ==================== BASIC STATISTICS ====================

print("\n========== BASIC STATISTICS ==========")

# Display basic statistical information
print(df.describe())


# ==================== DATA VISUALIZATION ====================

print("\n========== DATA VISUALIZATION ==========")


# 1. Age Distribution

print("\n1. Showing Age Distribution...")

plt.hist(df["Age"], bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()


# 2. Income Distribution

print("\n2. Showing Income Distribution...")

plt.hist(df["Income"], bins=10, color="orange")
plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Number of Customers")
plt.show()


# 3. Income and Account Balance

print("\n3. Showing Income and Account Balance...")

plt.scatter(
    df.index,
    df["Income"],
    color="red",
    label="Income"
)

plt.scatter(
    df.index,
    df["AccountBalance"],
    color="yellow",
    label="Account Balance"
)

plt.xlabel("Customer")
plt.ylabel("Amount")
plt.title("Income and Account Balance")
plt.legend()
plt.show()


# ==================== DATA ANALYSIS ====================

print("\n========== DATA ANALYSIS ==========")


# 4. Average Income by Gender

print("\n4. Calculating Average Income by Gender...")

avg_income = df.groupby("Gender")["Income"].mean()

print(avg_income)


# 5. Top 5 Customers

print("\n5. Finding Top 5 Customers by Account Balance...")

top_customers = df.nlargest(5, "AccountBalance")

print(top_customers[["Name", "AccountBalance"]])


print("\n========== ANALYSIS COMPLETED ==========")
```
